```python
import ipywidgets as widgets
from IPython.display import display, HTML
import matplotlib.pyplot as plt

from IPython.display import Audio

class PersonalFinanceApp:
  def __init__(self):
    self.create_widgets()

     # The speak method has been completely removed as requested.

  def create_widgets(self):
    """Sets up the main application UI elements."""
    self.logo_label = widgets.Label(value="Personal Finance Management System", style={'font_size': '50px', 'font_weight': 'bold'})
    display(self.logo_label)

    # Button definitions
    self.track_expenses_button = widgets.Button(description="Track Your Expenses", layout=widgets.Layout(width='300px'))
    self.smart_investments_button = widgets.Button(description="See Smart Investments", layout=widgets.Layout(width='300px'))
    self.money_tips_button = widgets.Button(description="See Money Tips", layout=widgets.Layout(width='300px'))
    self.Calculate_Debt_Mortgage = widgets.Button(description="Calculate Debt/Mortgage", layout=widgets.Layout(width='300px'))

    # Event handlers
    self.track_expenses_button.on_click(self.track_expenses)
    self.smart_investments_button.on_click(self.see_smart_investments)
    self.money_tips_button.on_click(self.see_money_tips)
    self.Calculate_Debt_Mortgage.on_click(self.calculate_debt_mortgage)

    # Display buttons
    display(self.track_expenses_button, self.smart_investments_button, self.Calculate_Debt_Mortgage, self.money_tips_button)

  def track_expenses(self, b):
    """Handles the expense tracking flow and displays a basic budget."""
    def on_submit(expenses):
      try:
        expenses = float(expenses)
        if expenses <= 2500:
          message = "Way to go! Check out the smart investments."
        else:
          # Budget categories and percentages (totaling 100%)
          categories = {"Rent": 0.4,
          "Utilities": 0.2,
          "Internet and Phone": 0.1,
          "Groceries": 0.15,
          "Transportation": 0.1,
          "Health": 0.05,
          "Personal Care": 0.04,
          "Entertainment": 0.05,
          "Emergency Fund": 0.1,
          "Other Savings": 0.04}

          message = "Here's your budget plan:\n\n"
          total_fixed_expenses = 0
          total_variable_expenses = 0
          total_savings = 0

          for category, percentage in categories.items():
            amount = expenses * percentage
            if category == "Emergency Fund" or category == "Other Savings":
              total_savings += amount
            else:
              if category in ["Rent", "Utilities", "Internet and Phone"]:
                total_fixed_expenses += amount
              else:
                total_variable_expenses += amount
              message += f"{category}: {amount:.2f} AED\n"

          message += f"\nTotal Fixed Expenses: {total_fixed_expenses:.2f} AED\n"
          message += f"Total Variable Expenses: {total_variable_expenses:.2f} AED\n"
          message += f"Total Savings: {total_savings:.2f} AED"

        html_message = message.replace('\n', '<br>')
        display(HTML(f"<p>{html_message}</p>"))
        # Removed: self.speak(message)

      except ValueError:
        display(HTML("<p>Please enter a valid number.</p>"))
        # Removed: self.speak("Please enter a valid number.")

    expenses_input = widgets.Text(description="Expenses (AED):")
    submit_button = widgets.Button(description="Submit")
    submit_button.on_click(lambda b: on_submit(expenses_input.value))
    display(expenses_input, submit_button)

  def see_smart_investments(self, b):
    """Displays a link to Google Finance for investment research."""
    website_url = "https://www.google.com/finance/?hl=en"
    display(HTML(f'<a href="{website_url}" target="_blank">Click here to see smart investments</a>'))
    # Removed: self.speak("Click here to see smart investments")

  def see_money_tips(self, b):
    """Calculates and displays money management tips based on a 50/30/20 rule."""
    def on_submit(income):
      try:
        income = float(income)
        percentage_savings = 0.20
        percentage_expenses = 0.50
        percentage_investments = 0.30

        savings_amount = income * percentage_savings
        expenses_amount = income * percentage_expenses
        investments_amount = income * percentage_investments

        message = "Here are some money management tips based on your income:\n\n"
        message += f"1. Savings (20%): Aim to save at least AED{savings_amount:.2f} per month. Consider creating an emergency fund.\n"
        message += f"2. Expenses (50%): Keep your monthly expenses below AED{expenses_amount:.2f}. Budget wisely and track your spending.\n"
        message += f"3. Investments (30%): Allocate AED{investments_amount:.2f} per month for investments. Consider diversifying your portfolio.\n"
        message += "4. Consider investing a portion of your savings in mutual funds for long-term growth.\n"
        message += "5. Keep an eye on your debt-to-income ratio. Try to minimize high-interest debt and pay off debts strategically.\n"
        message += "6. Review your financial goals regularly and adjust your budget and investment strategy accordingly.\n"


        html_message = message.replace('\n', '<br>')
        display(HTML(f"<p>{html_message}</p>"))
        # Removed: self.speak(message)

      except ValueError:
        display(HTML("<p>Please enter a valid number.</p>"))
        # Removed: self.speak("Please enter a valid number.")

    income_input = widgets.Text(description="Income (AED):")
    submit_button = widgets.Button(description="Submit")
    submit_button.on_click(lambda b: on_submit(income_input.value))
    display(income_input, submit_button)

  def calculate_debt_mortgage(self, b):
    self.option = widgets.RadioButtons(
       options=['Debt Repayment Plan', 'Mortgage Payment Calculation'],
       description='Choose:',
       style={'description_width': 'initial'}
    )
    self.principal_input = widgets.FloatText(description=' Principal (AED):')
    self.rate_input = widgets.FloatText(description=' Annual Interest Rate (%):')
    self.years_input = widgets.FloatText(description='Term (Years) :')

    self.submit_button = widgets.Button(description='Submit', button_style='success')
    self.submit_button.on_click(self.handle_debt_mortgage_submit)

    display(self.option, self.principal_input, self.rate_input, self.years_input, self.submit_button)

  def handle_debt_mortgage_submit(self, b):
    principal = self.principal_input.value
    r_annual = self.rate_input.value / 100
    n_years = self.years_input.value

    if self.option.value == 'Mortgage Payment Calculation':
      self.calculate_mortgage_payment(principal, r_annual, n_years)
    else:
      self.show_debt_plan(principal, r_annual, n_years)


  def calculate_mortgage_payment(self, principal, r_annual, n_years):
    r_monthly = r_annual / 12
    n_months = n_years * 12

    M = principal * (r_monthly * (1 + r_monthly)**n_months) / ((1 + r_monthly)**n_months - 1)
    total_paid = M * n_months
    total_interest = total_paid - principal

    display(HTML(f"""
      <h4>📘 Mortgage Payment Results:</h4>
      <p>Monthly Payment: <b>{M:,.2f} AED</b></p>
      <p>Total Interest Paid: <b>{total_interest:,.2f} AED</b></p>
    """))

  def show_debt_plan(self, principal, r_annual, n_years):
    r_monthly = r_annual / 12

    # Low payment – full term
    n_low = n_years * 12
    M_low = principal * (r_monthly * (1 + r_monthly)**n_low) / ((1 + r_monthly)**n_low - 1)

    # Medium payment – 80% of term
    n_med = n_low * 0.8
    M_med = principal * (r_monthly * (1 + r_monthly)**n_med) / ((1 + r_monthly)**n_med - 1)

    # Aggressive payment – 60% of term
    n_high = n_low * 0.6
    M_high = principal * (r_monthly * (1 + r_monthly)**n_high) / ((1 + r_monthly)**n_high - 1)


    display(HTML(f"""
      <h4>📊 Debt Repayment Options:</h4>
      <ul>
        <li>Low Payment Plan: <b>{M_low:,.2f} AED/month</b> (Full {n_years} years)</li>
        <li>Medium Payment Plan: <b>{M_med:,.2f} AED/month</b> (80% term)</li>
        <li>Aggressive Payment Plan: <b>{M_high:,.2f} AED/month</b> (60% term)</li>
      </ul>
    """))



# Instantiate and run the application
finance_app = PersonalFinanceApp()
```
